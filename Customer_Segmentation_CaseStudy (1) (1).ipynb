```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv("marketing_campaign.csv", sep="\t")

print(df.head())
print("Dataset shape:", df.shape)
print(df.info())
```

```python
# Fill missing income values
df["Income"] = df["Income"].fillna(df["Income"].median())

# Convert registration date into datetime
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    dayfirst=True,
    errors="coerce"
)

# Calculate customer tenure
latest_date = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (
    latest_date - df["Dt_Customer"]
).dt.days

# Remove unnecessary columns
df.drop(
    columns=[
        "ID",
        "Dt_Customer",
        "Z_CostContact",
        "Z_Revenue",
        "Response"
    ],
    inplace=True,
    errors="ignore"
)

print(df.head())
```

```python
# Identify numerical and categorical columns
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(include="object").columns

# Fill missing numerical values
df[num_cols] = df[num_cols].fillna(
    df[num_cols].median()
)

# Fill missing categorical values
for col in cat_cols:
    df[col] = df[col].fillna(
        df[col].mode()[0]
    )

# Convert categorical columns into numerical columns
df_encoded = pd.get_dummies(
    df,
    columns=cat_cols,
    dtype=int
)

print(df_encoded.head())
print("Encoded shape:", df_encoded.shape)
```

```python
# Create total spending
spending_cols = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

df_encoded["Total_Spending"] = df_encoded[
    spending_cols
].sum(axis=1)

# Create total purchase count
purchase_cols = [
    "NumDealsPurchases",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases"
]

df_encoded["Total_Purchases"] = df_encoded[
    purchase_cols
].sum(axis=1)

# Remove original columns used to calculate totals
df_encoded.drop(
    columns=spending_cols + purchase_cols,
    inplace=True,
    errors="ignore"
)

print(df_encoded.head())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(df_encoded)

print("Scaled data shape:", X_scaled.shape)
```
